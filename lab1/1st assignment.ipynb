```python
import requests
import pandas as pd
import time

URL = "https://api.open-meteo.com/v1/forecast"

areas = {
    "RR Nagar": (12.9250, 77.5190),
    "Indiranagar": (12.9784, 77.6408),
    "Hebbal": (13.0358, 77.5970),
    "Electronic City": (12.8452, 77.6602),
    "Whitefield": (12.9698, 77.7499)
}

all_weather = []

for area, (latitude, longitude) in areas.items():

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m,relative_humidity_2m,precipitation",
        "timezone": "Asia/Kolkata",
        "forecast_days": 7
    }

    try:
        r = requests.get(URL, params=params, timeout=20)
        r.raise_for_status()

        result = r.json()

        wx = pd.DataFrame(result["hourly"])


        wx["time"] = pd.to_datetime(wx["time"])
        wx["area"] = area

        
        all_weather.append(wx)

        print(area, ":", len(wx), "records")

    except Exception as e:
        print("Error for", area, ":", e)

   
    time.sleep(1)


weather_df = pd.concat(all_weather, ignore_index=True)


weather_df.to_csv("bengaluru_weather.csv", index=False)

print("Total records:", len(weather_df))
print("CSV saved as bengaluru_weather.csv")

weather_df.head()
```

```python
import requests
import pandas as pd
import time

URL = "https://air-quality-api.open-meteo.com/v1/air-quality"

areas = {
    "RR Nagar": (12.9250, 77.5190),
    "Indiranagar": (12.9784, 77.6408),
    "Hebbal": (13.0358, 77.5970),
    "Electronic City": (12.8452, 77.6602),
    "Whitefield": (12.9698, 77.7499)
}

all_air_quality = []

for area, (latitude, longitude) in areas.items():

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "pm2_5,pm10,carbon_monoxide",
        "timezone": "Asia/Kolkata",
        "forecast_days": 7
    }

    try:
        r = requests.get(URL, params=params, timeout=20)
        r.raise_for_status()

        result = r.json()

        aq = pd.DataFrame(result["hourly"])
   
        aq["time"] = pd.to_datetime(aq["time"])
 
        aq["area"] = area
        all_air_quality.append(aq)

        print(area, ":", len(aq), "records")

    except Exception as e:
        print("Error for", area, ":", e)

  
    time.sleep(1)


air_quality_df = pd.concat(all_air_quality, ignore_index=True)

air_quality_df.to_csv("bengaluru_air_quality.csv", index=False)

print("Total records:", len(air_quality_df))
print("CSV saved as bengaluru_air_quality.csv")

air_quality_df.head()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

URL = "https://books.toscrape.com/catalogue/page-{}.html"

all_books = []

for page in range(1, 51):

    try:
        url = URL.format(page)

        r = requests.get(url, timeout=20)
        r.raise_for_status()

        soup = BeautifulSoup(r.text, "html.parser")

        books = soup.select("article.product_pod")

        for book in books:

            # Extract title
            title = book.h3.a["title"]

            # Extract price
            price = book.select_one(".price_color").text.strip()

            # Extract rating
            rating = book.select_one("p.star-rating")["class"][1]

            all_books.append({
                "title": title,
                "price": price,
                "rating": rating
            })

        print("Page", page, ":", len(books), "books")

    except Exception as e:
        print("Error on page", page, ":", e)

    time.sleep(1)

# Convert to DataFrame
books_df = pd.DataFrame(all_books)

# Save as CSV
books_df.to_csv("books_data.csv", index=False)

print("Total books:", len(books_df))
print("CSV saved as books_data.csv")

books_df.head()
```
